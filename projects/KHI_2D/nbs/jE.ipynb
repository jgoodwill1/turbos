```python
import xarray as xr
from pathlib import Path

# ds.close()

results = []
kf = 4

t = [140000, 240000, 320000, 480000]

for i in range(4):
    print(f"Processing timestep {t[i]}")

    ds = run.load_step(t[i], ['J', 'Ji', 'Je', 'E'])
    # print(ds)

    J = kfilter(ds['J'], kf)
    Ji = kfilter(ds['Ji'], kf)
    Je = kfilter(ds['Je'], kf)
    E = kfilter(ds['E'], kf)

    ds['JiE'] = xr.dot(Ji, E, dims="c")
    ds['JeE'] = xr.dot(Je, E, dims="c")
    ds['JE']  = xr.dot(J , E, dims="c")
    ds_JE = xr.Dataset({
        'JiE': ds['JiE'],
        'JeE': ds['JeE'],
        'JE':  ds['JE'],
    })

    # Save normalized time as dimension coordinate
    ds_JE = ds_JE.expand_dims(
        time=[ds.attrs['time']]
    )

    # Save additional time information
    ds_JE = ds_JE.assign_coords(
        time_simulation=('time', [ds.attrs['time_simulation']]),
        step=('time', [ds.attrs['step']])
    )
    results.append(ds_JE)

# Combine all timesteps
ds_all = xr.concat(results, dim='time')

# Save as NetCDF
ds_all.attrs['kf'] = kf
path = Path("~/turbos/projects/KHI_2D/data/JE.nc").expanduser()
ds_all.to_netcdf(path, mode="w")
```

```python
import matplotlib.pyplot as plt
import xarray as xr

ds = xr.open_dataset("~/turbos/projects/KHI_2D/data/JE.nc")

variables = ["JiE", "JeE", "JE"]

labels = [
    r"$j_i \cdot E$",
    r"$j_e \cdot E$",
    r"$j \cdot E$",
]

vmag = [6e-5, 6e-5, 6e-5]

fig, ax = plt.subplots(
    3, 4,
    figsize=(16, 8),
    sharex=True,
    sharey=True,
    layout= None,
    gridspec_kw={
        "wspace": 0.02,
        "hspace": 0.02
    }
)

fig.subplots_adjust(
    left=0.07,
    right=0.90,
    bottom=0.08,
    top=0.95,
    wspace=0.02,
    hspace=0.02
)

for row, (var, label, vmax) in enumerate(zip(variables, labels, vmag)):

    for col in range(4):

        da = ds[var].isel(time=col)

        im = ax[row, col].imshow(
            da.values,
            cmap="seismic",
            origin="lower",
            interpolation="auto",
            extent=[
                ds.y.values[0]/10, ds.y.values[-1]/10,
                ds.x.values[0]/10, ds.x.values[-1]/10
            ],
            vmin=-vmax,
            vmax=vmax,
            aspect="equal"
        )

        # im = ax[row, col].pcolormesh(da.y, da.x, da, cmap = 'seismic', vmin = -vmax, vmax = vmax)

        # Time labels across top
        if row == 0:
            ax[row, col].set_title(
                rf"$t \omega_{{ci}} = {ds.time.values[col]:g}$",
                fontsize=13
            )

        # Spatial axes
        if row == 3:
            ax[row, col].set_xlabel(r"$y(d_i)$")

        if col == 0:
            ax[row, col].set_ylabel(r"$x(d_i)$" )


    exponent = int(np.floor(np.log10(vmax)))
    scale = 10**exponent

    cbar = fig.colorbar(
        im,
        ax=ax[row, :],
        fraction=0.005,
        pad=0.01,
        ticks=[-vmax, 0, vmax],
        format="%.1e"
    )

    cbar.ax.tick_params(labelsize=9, pad=2, length=2)
    cbar.ax.set_title(labels[row], fontsize=13, pad=10, x=2.5)

plt.savefig("/home/2480/turbos/projects/KHI_2D/fig/JE.png", dpi=300)
plt.show()
```
